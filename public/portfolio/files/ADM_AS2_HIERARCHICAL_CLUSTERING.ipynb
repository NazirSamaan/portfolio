```python
# HYPOTHESIS 1: Distinct groups by Income & Spending
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

df = pd.read_csv('Mall_Customers.csv')
X = df[['Annual Income (k$)', 'Spending Score (1-100)']].values

# Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Dendrogram
Z = linkage(X_scaled, method='ward')
plt.figure(figsize=(10, 6))
dendrogram(Z, truncate_mode='level', p=5)
plt.title('Dendrogram Hierarchical: Clustering (Income vs Spending)')
plt.xlabel('Customers')
plt.ylabel('Distance')
plt.show()

```

```python
# HYPOTHESIS 2: Age significantly affects the customers’ spending behaviour.

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram, linkage

# Load dataset
df = pd.read_csv('Mall_Customers.csv')

#  Filtering Age
df_filtered = df[(df['Age'] >= 35) & (df['Age'] <= 55)]

# Standardize Age + Spending Score
X_scaled = StandardScaler().fit_transform(df_filtered[['Age', 'Spending Score (1-100)']])

# Linkage matrix
Z = linkage(X_scaled, method='ward')

# Plot dendrogram
plt.figure(figsize=(20, 6))
dendrogram(
    Z,
    labels=df_filtered['Age'].values,
    leaf_rotation=90,
    leaf_font_size=7
)

plt.title('Dendrogram – Ages vs Spending Score')
plt.xlabel('Customer Age')
plt.ylabel('Spending Score (Distance)')
plt.tight_layout()
plt.show()
```

```python
# HYPOTHESIS 3: Gender plays a role in influencing spending score.
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram, linkage

# Load data
df = pd.read_csv('Mall_Customers.csv')

# Encode Gender and select features
df_encoded = df.copy()
df_encoded['Gender_Encoded'] = df_encoded['Gender'].map({'Male': 0, 'Female': 1})
X = df_encoded[['Gender_Encoded', 'Spending Score (1-100)']].values

# Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Perform hierarchical clustering
Z = linkage(X_scaled, method='ward')

# Plot dendrogram
plt.figure(figsize=(10, 6))
dendrogram(Z, truncate_mode='level', p=5)
plt.title('Dendrogram: Gender vs Spending Score')
plt.xlabel('Customers')
plt.ylabel('Distance')
plt.show()
```

```python
# HYPOTHESIS 4: A small cluster of high-income, high-spending customers exists (premium segment).
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster

# Load the dataset
df = pd.read_csv('Mall_Customers.csv')

# Select features: Annual Income and Spending Score
X = df[['Annual Income (k$)', 'Spending Score (1-100)']].values

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Perform hierarchical clustering using Ward linkage
Z = linkage(X_scaled, method='ward')

# Plot dendrogram
plt.figure(figsize=(12, 6))
dendrogram(Z, truncate_mode='level', p=5)
plt.title('Dendrogram: Premium Market Segment (Income vs Spending)')
plt.xlabel('Customers')
plt.ylabel('Distance')
plt.show()
```
