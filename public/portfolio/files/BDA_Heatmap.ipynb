```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("teen_phone_addiction_dataset.csv")

# Select relevant columns
df_subset = df[['Addiction_Level', 'Anxiety_Level', 'Depression_Level', 'Self_Esteem']]

# Calculate correlation matrix
correlation_matrix = df_subset.corr(numeric_only=True)

# Show correlation matrix
print(correlation_matrix.to_string(float_format="%.6f"))

# Set figure size
plt.figure(figsize = (10,8))


# Plot heatmap parameters
sns.heatmap(
    correlation_matrix,
    cmap = 'viridis',
    vmin = -1,
    vmax = 1,
    center = 0,
    annot=True,
    fmt=".2f",
    square=True,
    linewidths=.5
)

plt.title("Correlation Heatmap: Addiction and Mental Health Factors")
plt.show()
```
